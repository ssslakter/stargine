```python
HEIGHT = 100
WIDTH = 100
zoom = 200.0
centerX = -0.5
centerY = 0.0
MAX_ITERATIONS = 1000


pixels = [0] * HEIGHT * WIDTH

def map_to_color(iter_count: float):
    intensity = int(iter_count * 255)
    return (intensity << 16) | (intensity << 8) | intensity | (255 << 24)

for y in range(HEIGHT):
    for x in range(WIDTH):
        x0 = (x - float(WIDTH) / 2.0) / zoom + centerX
        y0 = (y - float(HEIGHT) / 2.0) / zoom + centerY
        zx = 0.0
        zy = 0.0
        iter = 0

        while zx * zx + zy * zy < 4.0 and iter < MAX_ITERATIONS:
            temp = zx * zx - zy * zy + x0
            zy = 2.0 * zx * zy + y0
            zx = temp
            iter += 1

        pixels[y * WIDTH + x] = map_to_color(iter / MAX_ITERATIONS)
```
